```python
import chainladder as cl
import pandas as pd
import matplotlib.pyplot as plt
```

```python
data = pd.read_csv(
    "https://raw.githubusercontent.com/casact/chainladder-python/master/chainladder/utils/data/clrd.csv"
)

data2 = data[data["AccidentYear"] >= 1000]
```

```python
triangle2 = cl.Triangle(
    data2,
    origin="AccidentYear",
    development="DevelopmentYear",
    index=["GRNAME"],
    columns=["CumPaidLoss"],
)
```

```python
model = cl.Chainladder().fit(triangle2)

triangle_pred = model.full_triangle_
```

```python
triangle2.sum().T.plot(
    marker=".",
    title="Rzeczywiste dane "
)

plt.show()
```

```python
triangle_pred.sum().T.plot(
    marker=".",
    title="Chain Ladder doszacowany"
)

plt.show()
```

```python
(
    triangle_pred.sum()
    - triangle2.sum()
).T.plot(
    marker=".",
    title="Co dopisał Chain Ladder"
)

plt.show()
```
