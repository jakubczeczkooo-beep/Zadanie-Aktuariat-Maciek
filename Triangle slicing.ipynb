```python
import chainladder as cl
```

```python
# Load data
clrd = cl.load_sample('clrd')

# wszystko razem
clrd = clrd.groupby('LOB').sum()
clrd
```

```python
# skumulowany reszta out

clrd = clrd['CumPaidLoss']
```

```python
import matplotlib.pyplot as plt

fig, ax = plt.subplots(1, 3, figsize=(15, 4))

for i, lob in enumerate(["medmal", "wkcomp", "comauto"]):
    tri = (
        cl.load_sample("clrd")
        .groupby("LOB")
        .sum()["CumPaidLoss"]
        .loc[lob]
    )
    tri = tri.iloc[:, :, 4:, :]
    tri.link_ratio.T.plot(
        marker="o",
        ax=ax[i],
        title=lob
    )

   

plt.tight_layout()
plt.show()


#Wniosek medmal ma największe wahania link ratios, więc jest najtrudniejszy do modelowania metodą Chain Ladder
```
